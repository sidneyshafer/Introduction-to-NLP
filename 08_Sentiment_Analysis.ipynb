# Sentiment Analysis

## Overview
Sentiment analysis (also called opinion mining) is the process of determining the emotional tone behind text - whether it's positive, negative, or neutral. It's one of the most widely used NLP applications in business and research.

## What is Sentiment Analysis?

### Types of Sentiment:
- **Positive:** Expressing favorable opinion or emotion ("great", "amazing", "love it")
- **Negative:** Expressing unfavorable opinion or emotion ("terrible", "hate", "awful")
- **Neutral:** Factual, without emotional coloring ("I went to the store")

### Sentiment Scores:
- **Polarity:** Ranges from -1.0 (most negative) to +1.0 (most positive)
- **Compound Score:** Overall sentiment combining all words
- **Subjectivity:** Measures how subjective vs. objective the text is (0 = objective, 1 = subjective)

## Why is Sentiment Analysis Important?

### Real-World Applications:
1. **Customer Feedback:** Analyze product reviews, social media mentions
2. **Brand Monitoring:** Track public sentiment about companies/products
3. **Market Research:** Understanding customer opinions and trends
4. **Customer Support:** Prioritize urgent negative feedback
5. **Political Analysis:** Gauge public opinion on policies/candidates
6. **Financial Trading:** Sentiment analysis of news for trading decisions

## Tools Covered in This Notebook

### 1. TextBlob
- **Pros:** Simple API, easy to use, returns polarity + subjectivity
- **Cons:** Less accurate, rule-based, doesn't handle negation well
- **Use Case:** Quick prototyping, simple sentiment scoring

### 2. VADER (Valence Aware Dictionary and sEntiment Reasoner)
- **Pros:** Excellent for social media, handles slang/emoticons/emojis, negation-aware
- **Cons:** Rule-based (not ML), less effective on formal text
- **Use Case:** Social media analysis, short informal text
- **Scores:** Positive, negative, neutral, compound

### Choosing the Right Tool:
- **TextBlob:** Simple tasks, formal text, quick prototyping
- **VADER:** Social media, reviews, informal text with slang/emojis
- **Transformers:** Most accurate, but slower (covered in next notebook)

```python
# Sentiment Analysis

sentence_1 = "i had a great time at the movie it was really funny"
sentence_2 = "i had a great time at the movie but the parking was terrible"
sentence_3 = "i had a great time at the movie but the parking wasn't great"
sentence_4 = "i went to see a movie"
```

## Example Sentences
Four sentences with varying sentiment polarity and complexity.

```python
from textblob import TextBlob
```

---

## Part 1: TextBlob Sentiment Analysis
TextBlob provides a simple API for sentiment analysis based on pattern matching.

```python
print(sentence_1)
sentiment_score_1 = TextBlob(sentence_1)
print(sentiment_score_1.sentiment.polarity)
```

**Result:** High positive polarity (around 0.6+) indicates strong positive sentiment.

## Sentence 1: Positive Sentiment
"great time" and "really funny" are clearly positive.

```python
print(sentence_2)
sentiment_score_2 = TextBlob(sentence_2)
print(sentiment_score_2.sentiment.polarity)
```

**Result:** Lower positive score - the negative aspect reduces overall sentiment, but positive still dominates.

## Sentence 2: Mixed Sentiment
"great time" (positive) BUT "parking was terrible" (negative).

```python
print(sentence_3)
sentiment_score_3 = TextBlob(sentence_3)
print(sentiment_score_3.sentiment.polarity)
```

**Result:** Negation ("wasn't great") is detected, resulting in a similar mixed sentiment score.

## Sentence 3: Negation Test
Similar to sentence 2, but uses "wasn't great" instead of "terrible".

```python
print(sentence_4)
sentiment_score_4 = TextBlob(sentence_4)
print(sentiment_score_4.sentiment.polarity)
```

**Result:** Near-zero polarity indicates neutral sentiment (no positive or negative words).

## Sentence 4: Neutral Statement
Factual statement with no emotional content.

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
```

---

## Part 2: VADER Sentiment Analysis
VADER is specifically tuned for social media text and provides more nuanced sentiment scores.

```python
vader_sentiment = SentimentIntensityAnalyzer()
```

Initialize the VADER sentiment analyzer.

```python
print(sentence_1)
print(vader_sentiment.polarity_scores(sentence_1))
```

**Result:** High compound score and high pos value confirm strong positive sentiment.

## VADER Scores Explanation
VADER returns 4 scores:
- **neg:** Proportion of negative sentiment (0-1)
- **neu:** Proportion of neutral sentiment (0-1)
- **pos:** Proportion of positive sentiment (0-1)
- **compound:** Overall sentiment (-1 to +1) - **this is the key score**

### Compound Score Interpretation:
- **≥ 0.05:** Positive sentiment
- **> -0.05 and < 0.05:** Neutral sentiment
- **≤ -0.05:** Negative sentiment

## Sentence 1 with VADER
Clearly positive sentence analyzed.

```python
print(sentence_2)
print(vader_sentiment.polarity_scores(sentence_2))
```

**Result:** Both pos and neg have values, but compound might lean slightly positive or negative depending on which dominates.

## Sentence 2 with VADER
Mixed sentiment: positive and negative aspects.

```python
print(sentence_3)
print(vader_sentiment.polarity_scores(sentence_3))
```

**Result:** VADER handles negation well - "wasn't great" is correctly identified as somewhat negative.

## Sentence 3 with VADER
Negation handling test.

```python
print(sentence_4)
print(vader_sentiment.polarity_scores(sentence_4))
```

**Result:** High neutral score and compound near zero indicates no emotional content.

## Comparison: TextBlob vs VADER

| Feature | TextBlob | VADER |
|---------|----------|-------|
| **Ease of Use** | Very simple | Simple |
| **Output** | Polarity + Subjectivity | 4 detailed scores |
| **Negation Handling** | Basic | Excellent |
| **Social Media** | Poor | Excellent |
| **Formal Text** | Good | Good |
| **Emoticons/Emojis** | No | Yes |
| **Speed** | Fast | Fast |
| **Accuracy** | Moderate | High (for short text) |

## When to Use Each:

### Use TextBlob when:
- Quick prototyping needed
- Simple positive/negative/neutral classification
- Analyzing formal, well-written text
- Need subjectivity scores

### Use VADER when:
- Analyzing social media posts, tweets, reviews
- Text contains slang, abbreviations, emoticons
- Need granular sentiment breakdown
- Negation is important ("not good", "isn't bad")

## Key Insights:
1. **Context matters:** "great" + "terrible" = mixed sentiment
2. **Negation is tricky:** "wasn't great" ≠ "terrible" but both negative
3. **Neutral text:** Factual statements score near zero
4. **Tool selection:** Choose based on text type and accuracy needs
