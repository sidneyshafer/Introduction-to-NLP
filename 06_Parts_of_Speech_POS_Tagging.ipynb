# Parts of Speech (POS) Tagging

## Overview
POS tagging is the process of labeling each word in a text with its appropriate part of speech (noun, verb, adjective, etc.). It's crucial for understanding sentence structure and meaning.

## What is POS Tagging?

### Common POS Tags:
- **NOUN:** Person, place, thing (dog, city, happiness)
- **VERB:** Action or state (run, is, think)
- **ADJ (Adjective):** Describes noun (beautiful, red, fast)
- **ADV (Adverb):** Modifies verb/adj (quickly, very, well)
- **PRON (Pronoun):** Replaces noun (he, she, it, they)
- **DET (Determiner):** Introduces noun (the, a, this, some)
- **ADP (Adposition):** Preposition (in, on, at, with)
- **CONJ (Conjunction):** Connects words (and, but, or)

## Why is POS Tagging Important?

### 1. Disambiguation
Same word, different meaning:
- "**book** a flight" (verb)
- "read a **book**" (noun)

### 2. Applications
- **Lemmatization:** Requires POS for accurate results
- **Named Entity Recognition:** Nouns often indicate entities
- **Information Extraction:** Extract specific grammatical patterns
- **Machine Translation:** Grammar varies across languages
- **Text-to-Speech:** Pronunciation depends on POS (record vs. record)
- **Question Answering:** Understanding query structure

````python
# Parts of speech (POS) tagging

import spacy
import pandas as pd
````

## Setup
Import spaCy and pandas for POS analysis.

````python
nlp = spacy.load("en_core_web_sm")
# if you are running this for the first time, or recieve an error "Can't find model 'en_core_web_sm'", 
# then please run the follwing in your terminal: python -m spacy download en_core_web_sm
````

Load the English language model. If you encounter an error, run:
```bash

```

````python
emma_ja = "emma woodhouse handsome clever and rich with a comfortable home and happy disposition seemed to unite some of the best blessings of existence and had lived nearly twentyone years in the world with very little to distress or vex her she was the youngest of the two daughters of a most affectionate indulgent father and had in consequence of her sisters marriage been mistress of his house from a very early period her mother had died too long ago for her to have more than an indistinct remembrance of her caresses and her place had been supplied by an excellent woman as governess who had fallen little short of a mother in affection sixteen years had miss taylor been in mr woodhouses family less as a governess than a friend very fond of both daughters but particularly of emma between them it was more the intimacy of sisters even before miss taylor had ceased to hold the nominal office of governess the mildness of her temper had hardly allowed her to impose any restraint and the shadow of authority being now long passed away they had been living together as friend and friend very mutually attached and emma doing just what she liked highly esteeming miss taylors judgment but directed chiefly by her own"
print(emma_ja)
````

## Example Text: Jane Austen's Emma
A passage from the opening of "Emma" - notice the complex sentence structure and varied vocabulary.

````python
spacy_doc = nlp(emma_ja)
````

## Process Text with spaCy
Create a spaCy Doc object that contains tokens with their linguistic annotations.

````python
pos_df = pd.DataFrame(columns=['token', 'pos_tag'])
````

## Extract POS Tags
Create a dataframe to store each token and its POS tag.

````python
for token in spacy_doc:
    pos_df = pd.concat([pos_df,
                       pd.DataFrame.from_records([{'token': token.text, 'pos_tag': token.pos_}])],
                      ignore_index=True)
````

````python
pos_df.head(15)
````

Observations:
- "emma", "woodhouse" → PROPN (proper nouns - names)
- "handsome", "clever" → ADJ (adjectives - descriptive)
- "and" → CCONJ (coordinating conjunction)
- "with" → ADP (adposition/preposition)

## View Sample POS Tags
Look at the first 15 tokens and their tags.

````python
pos_df_counts = pos_df.groupby(['token', 'pos_tag']).size().reset_index(name='counts').sort_values(by='counts', ascending=False)
````

## Word Frequency Analysis
Count how often each word appears with each POS tag.

````python
pos_df_counts.head(10)
````

This shows the most frequent token-POS combinations. Notice how function words (of, and, the) dominate.

````python
pos_df_poscounts = pos_df_counts.groupby(['pos_tag'])['token'].count().sort_values(ascending=False)
````

## POS Distribution
Count how many tokens belong to each part of speech.

````python
pos_df_poscounts.head(10)
````

This reveals the grammatical composition:
- **NOUN** is most common (narrative text focuses on people, places, things)
- **VERB** is second (actions and states)
- **ADJ** and **ADV** add descriptive detail
- Function words (DET, ADP, PRON) provide structure

````python
nouns = pos_df_counts[pos_df_counts.pos_tag == 'NOUN'][:10]
nouns.head(10)
````

The most frequent nouns tell us what the passage is about: family relationships (mother, father, daughters, sisters), authority figures (governess), and emotions (affection).
