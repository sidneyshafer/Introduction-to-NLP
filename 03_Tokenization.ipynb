# Tokenization

## Overview
Tokenization is the process of breaking down text into smaller units called "tokens". These tokens can be words, sentences, or even subwords. It's a fundamental step in NLP that transforms text into a format that algorithms can process.

## Why is Tokenization Important?
- **Text Analysis:** Enables counting words, analyzing frequency, and understanding structure
- **Machine Learning:** ML models need structured input; tokens provide that structure
- **Language Understanding:** Helps identify boundaries between words and sentences
- **Feature Extraction:** Tokens become features for NLP tasks

## Types of Tokenization

### 1. Sentence Tokenization
Splits text into individual sentences. Useful for:
- Document summarization
- Sentiment analysis per sentence
- Question answering systems

### 2. Word Tokenization
Splits text into individual words (and punctuation). Useful for:
- Text classification
- Named entity recognition
- Part-of-speech tagging

## NLTK's Tokenization Tools
NLTK provides robust tokenizers that handle:
- Contractions (e.g., "can't", "won't")
- Punctuation
- Abbreviations (Dr., Mr., etc.)
- Special cases

````python
# Tokenization

import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize, sent_tokenize
````

## Setup
Import NLTK's tokenization functions and download required data.

````python
sentences = "Her cat's name is Luna. Her dog's name is Max"
````

## Example 1: Sentence Tokenization
Split a paragraph into individual sentences using `sent_tokenize()`.

````python
sent_tokenize(sentences)
````

The text has been split into two sentences based on the period. NLTK's sentence tokenizer is smart enough to distinguish periods that end sentences from other uses (like abbreviations).

````python
sentence = "her cat's name is luna"
word_tokenize(sentence)
````

Notice how:
- Each word becomes a separate token
- The apostrophe in "cat's" is handled correctly
- Contractions are preserved

## Example 2: Word Tokenization (Simple)
Split a sentence into words using `word_tokenize()`.

````python
sentence_2 = "Her cat's name is Luna and her dog's name is Max"
word_tokenize(sentence_2)
````

Key observations:
- Possessive pronouns ("cat's", "dog's") are kept intact
- Capitalization is preserved (important for NER tasks)
- The word "and" (a stop word) is included - you can remove it later if needed

## Tokenization vs. Simple Split
You might wonder: why not just use `text.split()`?

**Simple split problems:**
```python
"Don't go!".split()  # Returns: ["Don't", "go!"]
```

**NLTK tokenization:**
```python
word_tokenize("Don't go!")  # Returns: ["Do", "n't", "go", "!"]
```

NLTK properly handles:
- Punctuation separation
- Contractions
- Special characters
- Multiple languages
