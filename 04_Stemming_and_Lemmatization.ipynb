# Stemming and Lemmatization

## Overview
Both stemming and lemmatization reduce words to their base or root form, but they use different approaches. This normalization helps group different forms of the same word together, reducing vocabulary size and improving model performance.

## Stemming vs. Lemmatization

### Stemming
- **Method:** Removes suffixes using heuristic rules (crude chopping)
- **Speed:** Fast
- **Output:** May not be a real word (e.g., "connect" → "connect", "connecting" → "connect")
- **Algorithm:** Porter Stemmer, Snowball Stemmer, Lancaster Stemmer
- **Use Case:** Information retrieval, search engines where speed matters

### Lemmatization
- **Method:** Uses vocabulary and morphological analysis
- **Speed:** Slower (requires dictionary lookup)
- **Output:** Always a real word (lemma)
- **Algorithm:** WordNet Lemmatizer
- **Use Case:** NLP tasks requiring grammatical accuracy

## When to Use Each?

**Use Stemming when:**
- Speed is critical
- Slight inaccuracies are acceptable
- Building search engines or information retrieval systems

**Use Lemmatization when:**
- Accuracy is important
- Working with formal text analysis
- Need readable output
- Doing semantic analysis or question answering

## Key Differences Example
| Original | Stemming | Lemmatization |
|----------|----------|---------------|
| running | run | running (without POS) / run (with POS) |
| better | better | good (as adjective) |
| feet | feet | foot |
| studies | studi | study |

```python
# Stemming

from nltk.stem import PorterStemmer
```

## Part 1: Stemming with Porter Stemmer
The Porter Stemmer is the most common stemming algorithm, using a series of rules to remove suffixes.

```python
ps = PorterStemmer()
```

Initialize the Porter Stemmer object.

```python
connect_tokens = ['connecting', 'connected', 'connectivity', 'connect', 'connects']
```

## Example 1: Stemming "Connect" Variations
Let's see how different forms of "connect" are reduced to their stem.

```python
for t in connect_tokens:
    print(t, ": ", ps.stem(t))
```

All variations are reduced to "connect" - perfect for grouping related words!

```python
learn_tokens = ['learned', 'learning', 'learn', 'learns', 'learner', 'learners']
```

## Example 2: Stemming "Learn" Variations

```python
for t in learn_tokens:
    print(t, ": ", ps.stem(t))
```

All "learn" variants are stemmed to "learn".

```python
likes_tokens = ['likes', 'better', 'worse']
```

## Example 3: Stemming Limitations
Here we see the limitations of stemming - it's just rule-based chopping.

```python
for t in likes_tokens:
    print(t, ": ", ps.stem(t))
```

Notice:
- "likes" → "like" (good!)
- "better" and "worse" remain unchanged (stemming doesn't understand semantics)
- Stemming can't reduce "better" to "good" or "worse" to "bad"

```python
# Lemmatization

import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

---

## Part 2: Lemmatization with WordNet
Lemmatization uses a dictionary (WordNet) to return the base or dictionary form of a word.

```python
lm = WordNetLemmatizer()
```

Initialize the WordNet Lemmatizer.

```python
for t in connect_tokens:
    print(t, ": ", lm.lemmatize(t))
```

Notice that without specifying part-of-speech (POS), lemmatization assumes words are nouns:
- "connecting" → "connecting" (kept as is, because as a noun it doesn't change)
- "connectivity" → "connectivity" (kept as is)

**Important:** For better lemmatization, you should provide POS tags (verb, noun, adjective, etc.).

## Comparing Lemmatization to Stemming
Let's compare how lemmatization handles the same "connect" words.

```python
for t in learn_tokens:
    print(t, ": ", lm.lemmatize(t))
```

Similar results with "learn" variations.

```python
for t in likes_tokens:
    print(t, ": ", lm.lemmatize(t))
```

Again, without POS context, lemmatization doesn't change comparative adjectives like "better" and "worse".

## Summary Comparison

| Aspect | Stemming | Lemmatization |
|--------|----------|---------------|
| **Speed** | ⚡ Fast | 🐌 Slower |
| **Accuracy** | 📊 Lower | ✅ Higher |
| **Output** | May not be real word | Always real word |
| **Context** | No context needed | Better with POS tags |
| **Use Case** | Search, IR | Text analysis, NLU |

## Best Practices
1. **For search engines:** Use stemming (speed matters more)
2. **For sentiment analysis:** Use lemmatization (context matters)
3. **For topic modeling:** Either works, but stemming is faster
4. **Always experiment:** Test both on your specific dataset
