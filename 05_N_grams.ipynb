# N-grams

## Overview
N-grams are contiguous sequences of N items (words, characters, or tokens) from a given text. They capture local word patterns and co-occurrences, providing context that individual words alone cannot convey.

## What are N-grams?

### Types of N-grams
- **Unigram (1-gram):** Single word - "the"
- **Bigram (2-gram):** Two consecutive words - "natural language"
- **Trigram (3-gram):** Three consecutive words - "natural language processing"
- **N-gram:** N consecutive words

### Example Breakdown
**Text:** "I love natural language processing"

- Unigrams: ["I", "love", "natural", "language", "processing"]
- Bigrams: ["I love", "love natural", "natural language", "language processing"]
- Trigrams: ["I love natural", "love natural language", "natural language processing"]

## Why Use N-grams?

### 1. Context Preservation
- Unigrams lose context: "not good" → ["not", "good"]
- Bigrams preserve meaning: "not good" → ["not good"]

### 2. Common Applications
- **Language Modeling:** Predicting next word (autocomplete, keyboards)
- **Feature Engineering:** ML features for text classification
- **Phrase Detection:** Finding common multi-word expressions
- **Text Generation:** Creating coherent text based on patterns
- **Plagiarism Detection:** Comparing text similarity

### 3. Trade-offs

| N-gram Size | Pros | Cons |
|-------------|------|------|
| Unigrams | Simple, low memory | No context |
| Bigrams | Good context/memory balance | May miss longer phrases |
| Trigrams+ | Rich context | Sparse, high memory, overfitting |

## Best Practices
- Start with bigrams and trigrams
- Remove low-frequency n-grams (noise reduction)
- Use with TF-IDF weighting for better importance scoring
- Consider skip-grams for non-contiguous patterns

```python
# N-grams

import nltk
import pandas as pd
import matplotlib.pyplot as plt
```

## Setup
Import required libraries for n-gram analysis and visualization.

```python
tokens = ['the', 'rise', 'of', 'artificial', 'intelligence', 'has', 'led', 'to', 'significant', 'advancements', 'in', 'natural', 'language', 'processing', 'computer', 'vision', 'and', 'other', 'fields', 'machine', 'learning', 'algorithms', 'are', 'becoming', 'more', 'sophisticated', 'enabling', 'computers', 'to', 'perform', 'complex', 'tasks', 'that', 'were', 'once', 'thought', 'to', 'be', 'the', 'exclusive', 'domain', 'of', 'humans', 'with', 'the', 'advent', 'of', 'deep', 'learning', 'neural', 'networks', 'have', 'become', 'even', 'more', 'powerful', 'capable', 'of', 'processing', 'vast', 'amounts', 'of', 'data', 'and', 'learning', 'from', 'it', 'in', 'ways', 'that', 'were', 'not', 'possible', 'before', 'as', 'a', 'result', 'ai', 'is', 'increasingly', 'being', 'used', 'in', 'a', 'wide', 'range', 'of', 'industries', 'from', 'healthcare', 'to', 'finance', 'to', 'transportation', 'and', 'its', 'impact', 'is', 'only', 'set', 'to', 'grow', 'in', 'the', 'years', 'to', 'come']
print(tokens)
```

## Sample Text: AI and Machine Learning
A tokenized passage about artificial intelligence and machine learning.

```python
unigrams = (pd.Series(nltk.ngrams(tokens, 1)).value_counts())
print(unigrams)
```

Notice which words appear most frequently:
- "of" and "to" are the most common (stop words!)
- Content words like "learning", "processing" appear less frequently
- This demonstrates why stop word removal is important

## Example 1: Unigrams (1-grams)
Individual words with their frequency counts.

```python
unigrams[:10].sort_values().plot.barh(color="lightsalmon", width=.9, figsize=(12,8))
plt.title("10 Most Frequently Occuring Unigrams")
```

## Visualization: Top 10 Unigrams
A horizontal bar chart showing the most frequent single words.

```python
bigrams = (pd.Series(nltk.ngrams(tokens, 2)).value_counts())
print(bigrams[:10])
```

Bigrams provide better context than unigrams:
- Phrases like "natural language", "machine learning" emerge
- "of" pairs show domain-specific concepts
- These patterns are useful for understanding document topics

## Example 2: Bigrams (2-grams)
Pairs of consecutive words reveal meaningful phrases.

```python
trigrams = (pd.Series(nltk.ngrams(tokens, 3)).value_counts())
print(trigrams[:10])
```

Trigrams show complete phrases:
- "natural language processing" (technical term)
- Longer phrases give more specific information
- But notice: trigrams are much sparser (fewer repetitions)

## Key Insights from This Analysis

### What We Learned:
1. **Unigrams** show word frequency but lack context
2. **Bigrams** reveal meaningful two-word phrases  
3. **Trigrams** provide rich context but are sparse

### Practical Tips:
- **Remove stop words** before generating n-grams for cleaner results
- **Bigrams** are often the sweet spot (context + manageable size)
- **Higher n-grams** (4+) can lead to data sparsity issues
- Combine n-grams of different sizes for comprehensive features

### Use in Machine Learning:
N-grams are used as features in:
- Text classification (spam detection, sentiment analysis)
- Information retrieval (search engines)
- Machine translation
- Spell checking and autocorrect
