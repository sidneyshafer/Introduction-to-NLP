# Text Vectorization (Bag of Words)

## Overview
Machine learning algorithms can't work with raw text - they need numbers. Text vectorization converts text into numerical representations that algorithms can process. The Bag of Words (BoW) model is one of the simplest and most common approaches.

## What is Bag of Words?

### Core Concept:
- Represent text as a "bag" (multiset) of words
- **Ignore** grammar, word order, and context
- **Count** how many times each word appears
- Each unique word becomes a feature (column)
- Each document becomes a vector (row)

### Example:
**Documents:**
1. "I love cats"
2. "I love dogs"
3. "I hate dogs"

**Vocabulary:** [I, love, cats, dogs, hate]

**BoW Matrix:**
```
      I  love  cats  dogs  hate
Doc1  1   1     1     0     0
Doc2  1   1     0     1     0
Doc3  1   0     0     1     1
```

## How CountVectorizer Works:

### Steps:
1. **Tokenization:** Split text into words
2. **Vocabulary Building:** Create list of all unique words across all documents
3. **Vectorization:** Count occurrences of each word in each document
4. **Matrix Creation:** Build document-term matrix

### Default Behavior:
- Converts to lowercase automatically
- Removes punctuation
- Tokenizes on whitespace
- Keeps all words (including stop words unless specified)

## Pros and Cons

### Advantages:
- ✅ Simple to understand and implement
- ✅ Fast to compute
- ✅ Works surprisingly well for many tasks
- ✅ Good for text classification, spam detection

### Disadvantages:
- ❌ Loses word order ("not good" = "good not")
- ❌ Ignores context and meaning
- ❌ Creates sparse, high-dimensional matrices
- ❌ No semantic understanding
- ❌ Can't capture phrases ("New York" → "New", "York")

## When to Use BoW:
- Document classification
- Spam detection
- Simple sentiment analysis
- Topic modeling (as input)
- When interpretability matters
- When you have limited data

````python
# Vectorizing Text

import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
````

## Setup
Import CountVectorizer from scikit-learn.

````python
data = [' Most shark attacks occur about 10 feet from the beach since that is where the people are',
        'the efficiency with which he paired the socks in the drawer was quite admirable',
        'carol drank the blood as if she were a vampire',
        'giving directions that the mountains are to the west only works when you can see them',
        'the sign said there was road work ahead so he decided to speed up',
        'the gruff old man sat in the back of the bait shop grumbling to himself as he scooped out a handful of worms']
````

## Sample Data
Six random sentences with varied vocabulary.

````python
countvec = CountVectorizer()
````

## Initialize CountVectorizer
Create a CountVectorizer object with default settings.

````python
countvec_fit = countvec.fit_transform(data)
````

## Fit and Transform
- `fit()` builds the vocabulary from the data
- `transform()` converts documents to vectors
- `fit_transform()` does both in one step

````python
bag_of_words = pd.DataFrame(countvec_fit.toarray(), columns = countvec.get_feature_names_out())
````

## Create DataFrame for Readability
Convert the sparse matrix to a pandas DataFrame to visualize the Bag of Words representation.

````python
print(bag_of_words)
````

### Reading the Matrix:
- Each **row** represents a document
- Each **column** represents a unique word from the vocabulary
- Each **value** is the count of that word in that document
- Most values are **0** (sparse matrix) - most words don't appear in most documents

### Key Observations:
1. **Common words** appear across multiple documents (e.g., "the", "was")
2. **Unique words help distinguish documents** (e.g., "shark", "vampire", "socks")
3. **Matrix is sparse:** Most cells are 0
4. **Word order is lost:** Can't tell if "the beach" or "beach the" was original

### Important CountVectorizer Parameters:
```python
CountVectorizer(
    max_features=100,      # Keep only top 100 most frequent words
    min_df=2,              # Ignore words appearing in < 2 documents
    max_df=0.8,            # Ignore words appearing in > 80% of documents
    ngram_range=(1,2),     # Include unigrams and bigrams
    stop_words='english'   # Remove English stop words
)
```

## Next Steps:
- **TF-IDF:** Weight words by importance, not just frequency
- **Word2Vec/GloVe:** Dense vectors capturing semantic meaning
- **Transformers:** Context-aware embeddings (BERT, GPT)

## Use Cases for Bag of Words:
1. **Text Classification:** Spam vs. ham, sentiment classification
2. **Document Similarity:** Compare documents based on word overlap
3. **Information Retrieval:** Simple search engines
4. **Feature Engineering:** Input for ML models (Naive Bayes, Logistic Regression)

## Limitations to Remember:
- "I love this, not hate it" and "I hate this, not love it" → Same vector!
- "New York" becomes separate "new" and "york" tokens
- High dimensionality (one column per unique word)
- No semantic understanding ("good" and "excellent" are completely different)
