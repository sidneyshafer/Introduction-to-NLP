# Text Preprocessing

## Overview
Text preprocessing is the foundational step in Natural Language Processing (NLP) that transforms raw text into a clean, standardized format suitable for analysis and machine learning models. This notebook covers essential preprocessing techniques.

## Key Concepts

### 1. Lowercasing
Converting all text to lowercase ensures consistency and reduces vocabulary size. For example, "Cat", "cat", and "CAT" are treated as the same word.

**Why it matters:**
- Reduces dimensionality
- Improves model generalization
- Prevents treating same words differently

### 2. Stop Words Removal
Stop words are common words (like "the", "is", "at", "which") that occur frequently but carry little meaningful information for many NLP tasks.

**NLTK Stop Words:**
- NLTK provides pre-built lists of stop words for multiple languages
- You can customize the list by adding or removing words based on your specific use case

### 3. When to Use Each Technique
- **Lowercasing:** Almost always useful, except for tasks where case matters (e.g., Named Entity Recognition)
- **Stop Word Removal:** Beneficial for topic modeling, text classification, but may hurt sentiment analysis (words like "not" are important)

```python
sentence = "Her cat's name is Luna"
```

## Example 1: Lowercasing a Single Sentence
We'll start with a simple example of converting text to lowercase.

```python
lower_sentance = sentence.lower()
```

```python
print(lower_sentance)
```

Notice how all characters are now lowercase, making the text more uniform.

```python
sentence_list = ['Could you pass me the TV remote?',
                'It is IMPOSSIBLE to find this hotel',
                 'Want to go for dinner on Tuesday?']
```

## Example 2: Lowercasing Multiple Sentences
Using list comprehension, we can efficiently process multiple sentences at once.

```python
lower_sentence_list = [x.lower() for x in sentence_list]
```

```python
print(lower_sentence_list)
```

All sentences are now standardized to lowercase, regardless of their original capitalization.

```python
# NLTK - Natural Language Toolkit
# Removing stop words - nltk has a predefined list of stopwords

import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

## Stop Words Removal with NLTK
NLTK (Natural Language Toolkit) provides a comprehensive list of stop words for English and other languages.

```python
en_stopwords = stopwords.words('english')
```

```python
print(en_stopwords)
```

The output shows 179 common English words that are typically filtered out during text preprocessing.

```python
sentence = 'it was too far to go to the shop and he did not want her to walk'
```

## Removing Stop Words from a Sentence
Let's apply stop word removal to clean a sentence.

```python
sentence_no_stopwords = ' '.join([word for word in sentence.split() if word not in en_stopwords])
```

```python
print(sentence_no_stopwords)
```

Notice how words like "it", "was", "to", "the", etc., have been removed, leaving only the meaningful content words.

```python
en_stopwords.remove("did")
en_stopwords.remove("not")
```

## Customizing the Stop Words List
You can modify the stop words list to suit your specific needs by removing or adding words.

```python
en_stopwords.append("go")
```

```python
sentence_no_stopwords = ' '.join([word for word in sentence.split() if word not in en_stopwords])
print(sentence_no_stopwords)
```

Now "did" and "not" are preserved (important for sentiment analysis), while "go" is removed as it may not be relevant for our analysis.
