# Regular Expressions (Regex) for NLP

## Overview
Regular expressions are powerful pattern-matching tools essential for text processing, cleaning, and information extraction in NLP tasks. They allow you to search, match, and manipulate text based on specific patterns.

## Key Concepts

### What are Regular Expressions?
A regular expression (regex) is a sequence of characters that defines a search pattern. Think of it as a mini-language for describing text patterns.

### Common Operations
1. **re.search()** - Finds the first occurrence of a pattern in a string
2. **re.sub()** - Substitutes/replaces patterns in a string
3. **re.findall()** - Finds all occurrences of a pattern

### Essential Regex Patterns

| Pattern | Description | Example |
|---------|-------------|---------|
| `?` | Makes preceding character optional | `sarah?` matches "sara" and "sarah" |
| `^` | Matches start of string | `^a` matches strings starting with "a" |
| `$` | Matches end of string | `y$` matches strings ending with "y" |
| `\|` | OR operator | `(need\|want)` matches "need" or "want" |
| `[^\w\s]` | Matches punctuation | Removes all non-word, non-space chars |
| `.` | Matches any character | `a.c` matches "abc", "a1c", etc. |
| `*` | 0 or more of preceding | `ab*` matches "a", "ab", "abb"... |
| `+` | 1 or more of preceding | `ab+` matches "ab", "abb"... |

### Use Cases in NLP
- Text cleaning (removing punctuation, special characters)
- Pattern extraction (emails, phone numbers, URLs)
- Name variations and spelling alternatives
- Data validation and formatting

### Raw Strings (r"...")
The `r` prefix creates a "raw string" which treats backslashes literally. This is crucial for regex and file paths in Windows.

```python
result_search = re.search("pattern", r"string to contain the pattern")
```

## Example 1: Using re.search()
`re.search()` returns a Match object if the pattern is found, or None if not found.

```python
print(result_search)
```

```python
result_search_2 = re.search("pattern", r"the phrase to find isn't in this string")
print(result_search_2)
```

The Match object shows where the pattern was found. If not found, it returns None.

```python
string = r"sara was able to help me find the items I needed quickly"
```

## Example 2: Using re.sub() for Substitution
Replace patterns in text - useful for correcting misspellings or standardizing names.

```python
new_string = re.sub("sara", "sarah", string)
print(new_string)
```

"sara" has been replaced with "sarah".

```python
customer_reviews = ['sam was a great help to me in the store', 
                    'the cashier was very rude to me, I think her name was eleanor', 
                    'amazing work from sadeen!', 
                    'sarah was able to help me find the items i needed quickly', 
                    'lucy is such a great addition to the team', 
                    'great service from sara she found me what i wanted']
```

## Example 3: Finding Reviews by Employee
Let's extract all reviews mentioning a specific employee (with spelling variations).

```python
sarahs_reviews = []
```

```python
pattern_to_find = r"sarah?"
```

### Pattern: sarah?
The `?` makes the 'h' optional, so this pattern matches both "sara" and "sarah".

```python
for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        sarahs_reviews.append(string)
```

```python
print(sarahs_reviews)
```

Successfully found all reviews mentioning "sara" or "sarah"!

```python
a_reviews = []
pattern_to_find = r"^a"

for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        a_reviews.append(string)
```

## Example 4: Anchor Patterns - Start of String (^)
The `^` anchor matches patterns at the **beginning** of a string.

```python
print(a_reviews)
```

Found reviews that start with the letter "a".

```python
y_reviews = []
pattern_to_find = r"y$"

for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        y_reviews.append(string)
```

## Example 5: Anchor Patterns - End of String ($)
The `$` anchor matches patterns at the **end** of a string.

```python
print(y_reviews)
```

Found reviews ending with the letter "y".

```python
need_want_reviews = []
pattern_to_find = r"(need|want)ed"

for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        need_want_reviews.append(string)
```

## Example 6: OR Operator (|)
Use parentheses and `|` to match multiple alternatives.

```python
print(need_want_reviews)
```

Found reviews containing either "needed" or "wanted".

```python
no_punct_reviews = []
pattern_to_find = r"[^\w\s]"
```

## Example 7: Removing Punctuation
Pattern `[^\w\s]` matches anything that is NOT a word character (\w) or whitespace (\s).

```python
for string in customer_reviews:
    no_punct_string = re.sub(pattern_to_find, "", string)
    no_punct_reviews.append(no_punct_string)
```

```python
print(no_punct_reviews)
```

All punctuation (commas, exclamation marks) has been removed, leaving only words and spaces.

## Key Takeaways
- Regular expressions are essential for text cleaning and pattern matching in NLP
- Use `?` for optional characters, `^` for start, `$` for end, `|` for OR
- `[^\w\s]` is perfect for removing punctuation
- Always test your patterns on sample data before applying to large datasets
