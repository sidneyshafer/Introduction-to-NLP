# Named Entity Recognition (NER)

## Overview
Named Entity Recognition is the task of identifying and classifying named entities (proper nouns) in text into predefined categories such as people, organizations, locations, dates, and more.

## What are Named Entities?

### Common Entity Types:
- **PERSON:** People's names (Larry Page, Sundar Pichai)
- **ORG:** Organizations, companies (Google, Stanford University)
- **GPE:** Geopolitical entities - countries, cities, states (California, California)
- **DATE:** Dates and time periods (September 4, 1998, October 24, 2015)
- **MONEY:** Monetary values ($100, 14%)
- **CARDINAL:** Numerals not in other types (14%, 56%)
- **ORDINAL:** Ordinal numbers (first, second)
- **LOC:** Non-GPE locations (mountain ranges, bodies of water)
- **PRODUCT:** Product names
- **EVENT:** Named events (wars, sports events)

## Why is NER Important?

### 1. Information Extraction
- Extract key facts from news articles
- Build knowledge graphs
- Populate databases automatically

### 2. Question Answering
- "Who founded Google?" → Identify PERSON entities
- "Where is Stanford University?" → Identify GPE/LOC entities

### 3. Content Recommendation
- Track mentions of companies, people, or products
- Recommend related articles based on entities

### 4. Anonymization
- Identify personal information for privacy protection
- Redact sensitive entity mentions

## spaCy's NER Capabilities
spaCy provides pre-trained NER models that:
- Recognize 18+ entity types
- Work across multiple languages
- Can be visualized with displaCy
- Support custom entity training

```python
import spacy
from spacy import displacy
from spacy import tokenizer
from IPython.display import HTML, display
import re
```

## Setup
Import spaCy, its visualization tool (displaCy), and other utilities.

```python
nlp = spacy.load("en_core_web_sm")
```

Load the pre-trained English model with NER capabilities.

```python
google_text = "Google was founded on September 4, 1998, by computer scientists Larry Page and Sergey Brin while they were PhD students at Stanford University in California. Together they own about 14% of its publicly listed shares and control 56% of its stockholder voting power through super-voting stock. The company went public via an initial public offering (IPO) in 2004. In 2015, Google was reorganized as a wholly owned subsidiary of Alphabet Inc. Google is Alphabet's largest subsidiary and is a holding company for Alphabet's internet properties and interests. Sundar Pichai was appointed CEO of Google on October 24, 2015, replacing Larry Page, who became the CEO of Alphabet. On December 3, 2019, Pichai also became the CEO of Alphabet."
print(google_text)
```

## Example Text: Google's History
A paragraph about Google containing various named entities (people, organizations, dates, locations).

```python
spacy_doc = nlp(google_text)
```

## Process Text with spaCy
Create a Doc object with NER annotations.

```python
for word in spacy_doc.ents:
    print(word.text, word.label_)
```

Entities detected:
- **ORG:** Google, Stanford University, Alphabet Inc
- **DATE:** September 4, 1998, 2004, 2015, October 24, 2015, December 3, 2019
- **PERSON:** Larry Page, Sergey Brin, Sundar Pichai
- **CARDINAL:** 14%, 56%
- **GPE:** California

Notice how spaCy correctly identifies:
- Organization names
- People's full names
- Specific dates
- Percentages and numbers
- Geographic locations

## Extract Named Entities
Loop through detected entities and print their text and type.

```python
html = displacy.render(spacy_doc, style="ent", jupyter=False)
display(HTML(html))
```

The visualization highlights each entity type with a different color, making it easy to see what spaCy detected.

## Visualize Entities with displaCy
Display entities with color-coded highlighting in the notebook.

```python
google_text_clean = re.sub(r'[^\W\S]', '', google_text).lower()
print(google_text_clean)
```

---

## Impact of Preprocessing on NER
Let's see what happens when we preprocess the text (lowercase + remove punctuation).

```python
spacy_doc_clean = nlp(google_text_clean)
```

Process the cleaned (lowercase) text.

```python
for word in spacy_doc_clean.ents:
    print(word.text, word.label_)
```

**Important Observation:**
NER performance **degrades significantly** with lowercased text! 

The model is trained on properly capitalized text and relies heavily on capitalization cues:
- Proper nouns start with capital letters
- Lowercase text loses this crucial signal
- Many entities may go undetected or be misclassified

### Lesson: Don't preprocess before NER!
Unlike other NLP tasks, NER requires:
- ✅ Preserve capitalization
- ✅ Keep punctuation (helps with sentence boundaries)
- ✅ Maintain original formatting
- ❌ Don't lowercase
- ❌ Don't remove too much punctuation

## Entities from Preprocessed Text

```python
html = displacy.render(spacy_doc_clean, style="ent", jupyter=False)
display(HTML(html))
```

## Summary

### Key Takeaways:
1. **NER extracts structured information** from unstructured text
2. **Capitalization is crucial** for accurate NER
3. **Pre-trained models** (like spaCy) work well out-of-the-box
4. **displaCy visualization** helps validate and understand NER results

### Best Practices:
- Use original, properly formatted text for NER
- Validate entity predictions (models aren't perfect)
- Consider domain-specific training for specialized texts (medical, legal, etc.)
- Combine NER with other NLP tasks (POS tagging, dependency parsing)

### Common Applications:
- **News analysis:** Track companies, people, events over time
- **Customer feedback:** Identify product and competitor mentions
- **Resume parsing:** Extract names, companies, skills, dates
- **Medical records:** Identify diseases, treatments, medications
- **Legal documents:** Extract party names, dates, monetary amounts
