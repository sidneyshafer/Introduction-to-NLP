# Transformer Models for NLP

## Overview
Transformer models represent a revolutionary advancement in NLP. Based on the "Attention is All You Need" paper (2017), transformers have become the foundation for state-of-the-art models like BERT, GPT, and T5.

## What are Transformers?

### Key Innovation: Attention Mechanism
- **Self-Attention:** Models can weigh the importance of different words in context
- **Parallel Processing:** Unlike RNNs, transformers process all words simultaneously
- **Long-Range Dependencies:** Captures relationships between distant words effectively

### Why Transformers Changed Everything:
1. **Better Context Understanding:** Self-attention captures nuanced relationships
2. **Transfer Learning:** Pre-trained models can be fine-tuned for specific tasks
3. **Scalability:** Train on massive datasets, then adapt to smaller tasks
4. **Multilingual:** Many models work across multiple languages

## Popular Transformer Models:

### BERT (Bidirectional Encoder Representations from Transformers)
- Reads text bidirectionally (left-to-right AND right-to-left)
- Excellent for understanding and classification tasks
- Use cases: Question answering, NER, text classification

### GPT (Generative Pre-trained Transformer)
- Unidirectional (left-to-right)
- Excellent for text generation
- Use cases: Creative writing, chatbots, code generation

### RoBERTa, DistilBERT, ALBERT
- Improvements/optimizations of BERT
- Trade-offs between accuracy, speed, and model size

## Hugging Face Transformers Library
The `transformers` library provides:
- Pre-trained models for dozens of tasks

- Support for PyTorch and TensorFlow

Import the transformers library and its pipeline functionality.

```python
sentence_1 = "i had a great time at the movie it was really funny"
sentence_2 = "i had a great time at the movie but the parking was terrible"
sentence_3 = "i had a great time at the movie but the parking wasn't great"
sentence_4 = "i went to see a movie"
```

## Test Sentences
Same sentences from the previous sentiment analysis notebook for comparison.

```python
sentiment_pipeline = pipeline("sentiment-analysis")
```

**Note:** The first time you run this, it will download the model (can take a few minutes). Subsequent runs will be much faster.

---

## Part 1: Default Sentiment Analysis Pipeline
Create a sentiment analysis pipeline using the default pre-trained model (DistilBERT).

```python
print(sentence_1)
sentiment_pipeline(sentence_1)
```

**Result:** POSITIVE with high confidence score (>0.99). Transformers provide a label and confidence score.

## Analyze Sentence 1
Clearly positive sentence with strong positive words.

```python
print(sentence_2)
sentiment_pipeline(sentence_2)
```

**Result:** The model must make a binary choice (POSITIVE or NEGATIVE). The confidence score tells us how certain it is.

## Analyze Sentence 2
Mixed sentiment: positive start, negative ending.

```python
print(sentence_3)
sentiment_pipeline(sentence_3)
```

**Result:** Transformers handle negation exceptionally well - they understand context, not just keywords.

## Analyze Sentence 3
Negation test: "wasn't great" should be negative.

```python
print(sentence_4)
sentiment_pipeline(sentence_4)
```

**Result:** Even neutral text gets classified as POSITIVE or NEGATIVE. The low confidence score indicates uncertainty.

## Analyze Sentence 4
Neutral, factual statement.

```python
specific_model = pipeline("sentiment-analysis", model="finiteautomata/bertweet-base-sentiment-analysis")
```

This model provides 3 labels: POS, NEG, NEU (positive, negative, neutral) - more nuanced than binary classification!

---

## Part 2: Using a Specific Model
Let's use a different model specifically trained for sentiment analysis: BERTweet (trained on Twitter data).

```python
print(sentence_1)
specific_model(sentence_1)
```

**Result:** Labeled as "POS" (positive) with high confidence.

## Test Sentence 1 with BERTweet

```python
print(sentence_2)
specific_model(sentence_2)
```

**Result:** Different models may classify mixed sentiment differently based on their training data.

## Test Sentence 2 with BERTweet
Mixed sentiment scenario.

```python
print(sentence_3)
specific_model(sentence_3)
```

## Test Sentence 3 with BERTweet

```python
print(sentence_4)
specific_model(sentence_4)
```

**Result:** BERTweet can classify as "NEU" (neutral), which is more appropriate for factual statements!

## Comparison: Traditional vs. Transformer Models

| Aspect | TextBlob/VADER | Transformers |
|--------|----------------|--------------|
| **Accuracy** | Moderate | Very High |
| **Context Understanding** | Limited | Excellent |
| **Speed** | Fast | Slower (but acceptable) |
| **Setup** | Simple | Requires model download |
| **Negation Handling** | Basic/Good | Excellent |
| **Model Size** | Tiny | Large (100MB-1GB+) |
| **Nuance** | Limited | Understands subtlety |

## Advantages of Transformers:
1. **Context-Aware:** Understand word relationships across entire sentence
2. **Pre-trained:** Trained on billions of words, capture language patterns
3. **Fine-tunable:** Can be adapted to specific domains
4. **Multilingual:** Many models work in 100+ languages
5. **State-of-the-art:** Best results for most NLP tasks

## When to Use Transformers:
- ✅ Accuracy is critical
- ✅ Have sufficient compute resources
- ✅ Complex text with nuance, sarcasm, or subtle sentiment
- ✅ Multi-sentence or paragraph-level analysis
- ❌ Don't use for: Real-time applications with strict latency requirements
