# Building a Text Classifier

## Overview
This notebook demonstrates how to build an **end-to-end text classification pipeline** using machine learning. We'll train multiple models to classify text and compare their performance.

## What is Text Classification?

### Definition:
Automatically categorizing text documents into predefined categories or labels.

### Real-World Examples:

| Application | Input | Output Categories |
|-------------|-------|-------------------|
| **Spam Detection** | Email text | Spam / Not Spam |
| **Sentiment Analysis** | Product review | Positive / Negative / Neutral |
| **Customer Support** | Support ticket | Technical / Billing / Account |
| **News Categorization** | Article text | Politics / Sports / Technology |
| **Language Detection** | Text string | English / Spanish / French |

## The ML Pipeline:

### Step-by-Step Process:
1. **Load Data:** Import labeled text data
2. **Preprocess Text:** Clean and normalize text
3. **Vectorization:** Convert text to numerical features (BoW, TF-IDF)
4. **Split Data:** Training set (learn patterns) and test set (evaluate)
5. **Train Models:** Fit classifiers on training data
6. **Evaluate:** Measure performance on unseen test data
7. **Compare:** Choose the best model

## Algorithms We'll Use:

### 1. Logistic Regression
- **Type:** Linear model
- **Strengths:** Fast, interpretable, good baseline
- **How it works:** Calculates probabilities using weighted sum of features

### 2. Naive Bayes (MultinomialNB)
- **Type:** Probabilistic model
- **Strengths:** Extremely fast, works well with text
- **How it works:** Uses Bayes' theorem with naive independence assumption
- **Why "Naive"?** Assumes features (words) are independent

### 3. Support Vector Machine (SVM)
- **Type:** Margin-based classifier
- **Strengths:** Powerful, handles high-dimensional data well
- **How it works:** Finds optimal decision boundary with maximum margin

## Evaluation Metrics:

### Accuracy:
$$\text{Accuracy} = \frac{\text{Correct Predictions}}{\text{Total Predictions}}$$

### Precision:
$$\text{Precision} = \frac{\text{True Positives}}{\text{True Positives} + \text{False Positives}}$$
*"Of all items predicted positive, how many are actually positive?"*

### Recall:
$$\text{Recall} = \frac{\text{True Positives}}{\text{True Positives} + \text{False Negatives}}$$
*"Of all actual positive items, how many did we identify?"*

### F1-Score:
$$\text{F1} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$$
*"Harmonic mean of precision and recall"*

## Confusion Matrix:
Visual representation of classification performance:

```
                Predicted
                Pos   Neg
Actual  Pos     TP    FN
        Neg     FP    TN
```
- **TP (True Positive):** Correctly predicted positive
- **TN (True Negative):** Correctly predicted negative
- **FP (False Positive):** Incorrectly predicted positive (Type I error)
- **FN (False Negative):** Incorrectly predicted negative (Type II error)

## Model 2: Naive Bayes

Probabilistic classifier that's extremely fast and works remarkably well for text.

**Why Naive Bayes for Text?**
- Extremely fast training and prediction
- Works well even with small datasets
- Effective with high-dimensional sparse data (like text)
- Based on Bayes' theorem: P(Class|Text) ∝ P(Text|Class) × P(Class)

## Key Takeaways:

### Building a Text Classifier:
1. **Load & Explore Data:** Understand distribution and balance
2. **Vectorize Text:** Convert to numerical features (BoW or TF-IDF)
3. **Split Data:** Train/test split
4. **Train Multiple Models:** Compare different algorithms
5. **Evaluate:** Use appropriate metrics (accuracy, precision, recall, F1)
6. **Compare:** Choose based on performance and requirements

### Choosing the Right Model:

**Use Logistic Regression when:**
- ✅ You need a quick, interpretable baseline
- ✅ You want to understand feature importance
- ✅ Speed and simplicity matter

**Use Naive Bayes when:**
- ✅ You have a small dataset
- ✅ Speed is critical (real-time prediction)
- ✅ Text classification is the task (it's optimized for it)

**Use SVM when:**
- ✅ You need maximum accuracy
- ✅ You have computational resources for training
- ✅ Interpretability is less important

### Common Improvements:

1. **Better Preprocessing:**
   - Remove stop words
   - Lemmatization/stemming
   - Handle negations ("not good" vs "good")

2. **Better Features:**
   - Use TF-IDF instead of BoW
   - Include bigrams/trigrams
   - Feature engineering (text length, punctuation, etc.)

3. **Hyperparameter Tuning:**
   - Grid search or random search
   - Cross-validation
   - Optimize C, kernel for SVM; alpha for Naive Bayes

4. **Handle Imbalanced Data:**
   - Oversample minority class (SMOTE)
   - Undersample majority class
   - Use class weights

5. **Ensemble Methods:**
   - Voting classifier combining multiple models
   - Stacking different classifiers
   - Random Forest or Gradient Boosting

### Metrics Selection Guide:

**Balanced Dataset:**
- Use **Accuracy** as primary metric

**Imbalanced Dataset:**
- Use **F1-Score** (balanced) or **Precision/Recall** (depending on goal)

**Cost-Sensitive Application:**
- **High Precision:** When false positives are costly (e.g., spam filter)
- **High Recall:** When false negatives are costly (e.g., medical diagnosis)

````python
countvec = CountVectorizer()
````

````python
countvec_fit = countvec.fit_transform(x)
````

````python
bag_of_words = pd.DataFrame(countvec_fit.toarray(), columns=countvec.get_feature_names_out())
````

````python
bag_of_words
````

````python
X_train, X_test, y_train, y_test = train_test_split(bag_of_words, y, test_size=0.3, random_state=7)
````

````python
lr = LogisticRegression(random_state=1).fit(X_train, y_train)
````

````python
y_pred_lr = lr.predict(X_test)
````

````python
accuracy_score(y_pred_lr, y_test)
````

````python
print(classification_report(y_test, y_pred_lr, zero_division=0))
````

````python
from sklearn.naive_bayes import MultinomialNB
````

````python
nb = MultinomialNB().fit(X_train, y_train)
````

````python
y_pred_nb = nb.predict(X_test)
````

````python
accuracy_score(y_pred_nb, y_test)
````

````python
print(classification_report(y_test, y_pred_nb, zero_division=0))
````

````python
from sklearn.linear_model import SGDClassifier
````

````python
svm = SGDClassifier().fit(X_train, y_train)
````

````python
y_pred_svm = svm.predict(X_test)
````

````python
accuracy_score(y_pred_svm, y_test)
````

````python
print(classification_report(y_test, y_pred_svm, zero_division=0))
````
